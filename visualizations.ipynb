```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import time

print("Libs Imported")
```

```python
def load_analysis_data(input_folder="processed_data"):
    loaded_dataframes = {}
    
    # Loop through all files in the folder
    for item in os.listdir(input_folder):
        if item.endswith(".parquet"):
            # Strip the .parquet extension to use as the dictionary key
            name = item.replace(".parquet", "")
            file_path = f"{input_folder}/{item}"
            
            # Read the parquet folder into a Pandas DataFrame
            loaded_dataframes[name] = pd.read_parquet(file_path)
            print(f"Successfully loaded: {name}")
            
    return loaded_dataframes

# 2. Execute the function
viz_data = load_analysis_data()
```

```python
sns.set_theme(style = "whitegrid")

df_best_sub_cat = viz_data['best_selling_sub_category']
plt.figure(figsize=(12, 6))

sns.barplot(
    data = df_best_sub_cat,
    x='sub-category',
    y='count',
    palette = 'viridis',
    legend = False 
)

plt.title('Best-selling Product Sub-categories', fontsize=16, fontweight='bold')
plt.xlabel('Sub-category', fontsize=12)
plt.ylabel('Order_Count', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style = "whitegrid")

highest_revenue = viz_data['highest_revenue']
plt.figure(figsize=(12, 6))

sns.barplot(
    data = highest_revenue,
    x='category',
    y='total_revenue',
    palette = 'viridis',
    legend = False 
)

plt.title('Category with Highest Revenue', fontsize=16, fontweight='bold')
plt.xlabel('Category', fontsize=12)
plt.ylabel('Revenue', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(style='plain', axis='y')
ax = plt.gca()
ax.get_yaxis().set_major_formatter(plt.FuncFormatter(lambda x, loc: "{:,}".format(int(x))))
plt.tight_layout()
plt.show()
```
