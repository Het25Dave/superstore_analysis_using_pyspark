```python
import pandas as pd 
import numpy as np
import matplotlib as plt
import os
import time

print("Libs Imported")
```

```python
def load_analysis_data(input_folder="processed_data"):
    loaded_dataframes = {}
    
    # Loop through all files in the folder
    for item in os.listdir(input_folder):
        if item.endswith(".parquet"):
            # Strip the .parquet extension to use as the dictionary key
            name = item.replace(".parquet", "")
            file_path = f"{input_folder}/{item}"
            
            # Read the parquet folder into a Pandas DataFrame
            loaded_dataframes[name] = pd.read_parquet(file_path)
            print(f"Successfully loaded: {name}")
            
    return loaded_dataframes

# 2. Execute the function
viz_data = load_analysis_data()
```

```python
df_best_sub_cat = viz_data['best_selling_sub_category']
df_best_sub_cat.plot(
    kind='bar',
    x='sub-category',
    y='count',
    legend=False,
    figsize=(12, 6),
    title='Best-selling Product Sub-categories'
)
plt.pyplot.xlabel('Sub-category')
plt.pyplot.ylabel('Order Count')
plt.pyplot.xticks(rotation=45, ha='right')
plt.pyplot.tight_layout()
plt.pyplot.show()
```
